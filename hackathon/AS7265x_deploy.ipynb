```python
import serial
print(serial.__file__)
import joblib
import numpy as np
import sklearn
print(sklearn.__version__)
```

```python
# Load the Random Forest model
model = joblib.load("random_forest_model.joblib")

# Set up serial communication (adjust the port to match your system)
arduino = serial.Serial('COM3', 9600)  # Change 'COM3' to your Arduino's port
arduino.flush()

def get_arduino_data():
    """Reads data from Arduino via Serial"""
    if arduino.in_waiting > 0:
        data = arduino.readline().decode('utf-8').strip()
        return data
    return None

def predict(sensor_data):
    """Predict using the trained Random Forest model"""
    # Assuming the model requires a list/array of 18 sensor readings
    prediction = model.predict([sensor_data])
    return prediction

# Main loop: continuously listen for button press and predict
while True:
    data = get_arduino_data()  # Get the sensor data line from Arduino
    if data:
        print(f"Received data: {data}")
        
        # Try parsing only if it looks like a CSV line with numbers
        if ',' in data and len(data.split(',')) == 18:
            try:
                sensor_values = list(map(float, data.split(',')))  # Convert string to floats
                print("Predicting...")
                prediction = predict(sensor_values)
                print(f"Prediction: {prediction}")
            except ValueError:
                print("Received malformed numeric data. Skipping.")
        else:
            print("Non-data message received. Skipping.")
```
